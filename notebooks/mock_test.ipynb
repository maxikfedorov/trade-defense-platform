```python
import requests
import json
from pprint import pprint

BASE_URL = "http://localhost:8000"


response = requests.get(f"{BASE_URL}/health")
print(f"GET /health")
print(f"Input: None")
print(f"Output: dict")
print(f"Status: {response.status_code}")
pprint(response.json())


print(f"POST /analyze")
print(f"Input: {{tnved_code: str, product_name?: str}}")
print(f"Output: {{metadata: dict, product_info: dict, metrics: dict, tariff_measures: dict, nontariff_measures: dict}}")

analyze_request = {
    "tnved_code": "8428 10"
}

response = requests.post(f"{BASE_URL}/analyze", json=analyze_request)
analysis_result = response.json()
print(f"Status: {response.status_code}")
pprint(analysis_result)

print(f"POST /generate")
print(f"Input: {{algorithm_result: dict, user_prompt?: str}}")
print(f"Output: {{rag_queries: list[str], sources_found: int, top_sources: list[dict], explanation: str}}")

generate_request = {
    "algorithm_result": analysis_result,
    "user_prompt": "Подробное объяснение"
}

response = requests.post(f"{BASE_URL}/generate", json=generate_request)
generation_result = response.json()
print(f"Status: {response.status_code}")
pprint(generation_result)


print(f"POST /full-pipeline")
print(f"Input: {{tnved_code: str, product_name?: str, user_prompt?: str}}")
print(f"Output: {{algorithm_result: dict, rag_queries: list[str], sources_found: int, top_sources: list[dict], explanation: str}}")

full_request = {
    "tnved_code": "8428 10",
    "product_name": "Лифты",
    "user_prompt": "Краткое резюме"
}

response = requests.post(f"{BASE_URL}/full-pipeline", json=full_request)
full_result = response.json()
print(f"Status: {response.status_code}")
pprint(full_result)


print(f"GET /config")
print(f"Input: None")
print(f"Output: dict")

response = requests.get(f"{BASE_URL}/config")
print(f"Status: {response.status_code}")
pprint(response.json())
```
