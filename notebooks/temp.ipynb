```python
# для файла data\postgres\тнвэд_список.xlsx

import pandas as pd
import json
import re

# Загрузка файла
file_path = r'data\postgres\тнвэд_список.xlsx'

# Читаем лист с соответствиями
sheet_name = 'ТН ВЭД - ОКПД2'  # Это тот самый лист, который на скриншоте!
print(f"Обрабатываем лист: {sheet_name}\n")

# Читаем с первой строкой как заголовок
df = pd.read_excel(file_path, sheet_name=sheet_name, header=0)

print("Структура таблицы:")
print(df.head(25))
print(f"\nВсего строк: {len(df)}")
print(f"Колонки: {df.columns.tolist()}\n")

# Функция очистки
def clean_value(val):
    if pd.isna(val) or str(val).strip() in ['', 'nan', '-', '--', 'NaN']:
        return None
    return str(val).strip()

# Парсинг данных
result = []
current_tnved = None
current_tnved_name = None

for idx, row in df.iterrows():
    # Получаем значения из колонок
    col_a = clean_value(row.iloc[0]) if len(row) > 0 else None  # Код ТН ВЭД
    col_b = clean_value(row.iloc[1]) if len(row) > 1 else None  # Наименование ТН ВЭД
    col_c = clean_value(row.iloc[2]) if len(row) > 2 else None  # Код ОКПД2
    col_d = clean_value(row.iloc[3]) if len(row) > 3 else None  # Наименование ОКПД2
    
    # Проверяем формат кода ТН ВЭД (XXXX XX или XXXX.XX)
    if col_a and re.search(r'\d{4}\s*\d{2}', col_a):
        tnved_code = re.sub(r'(\d{4})\s*(\d{2})', r'\1 \2', col_a)
        current_tnved = tnved_code
        current_tnved_name = col_b
    
    # Проверяем формат кода ОКПД2 (XX.XX.XX)
    if col_c and re.search(r'\d{2}\.\d{2}\.\d{2}', col_c):
        okpd_code = col_c
        okpd_name = col_d
        
        # Создаём запись с текущим кодом ТН ВЭД и найденным ОКПД
        result.append({
            'tnved_code': current_tnved,
            'tnved_name': current_tnved_name,
            'okpd_code': okpd_code,
            'okpd_name': okpd_name
        })

# Сохранение в JSON
output_file = 'tnved_okpd_mapping.json'
with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(result, f, ensure_ascii=False, indent=2)

print(f"{'='*80}")
print(f"✓ Успешно обработано записей: {len(result)}")
print(f"✓ Файл сохранён: {output_file}")

# Статистика
tnved_count = len(set(r['tnved_code'] for r in result if r['tnved_code']))
okpd_count = len(set(r['okpd_code'] for r in result if r['okpd_code']))

print(f"\nСтатистика:")
print(f"  - Уникальных кодов ТН ВЭД: {tnved_count}")
print(f"  - Уникальных кодов ОКПД: {okpd_count}")
print(f"  - Всего соответствий: {len(result)}")

# Примеры
print(f"\n{'='*80}")
print("Примеры записей:\n")
for i, record in enumerate(result[:5], 1):
    print(f"{i}. ТН ВЭД: {record['tnved_code']}")
    tnved_display = record['tnved_name'][:70] + '...' if record['tnved_name'] and len(record['tnved_name']) > 70 else record['tnved_name']
    print(f"   └─ {tnved_display}")
    print(f"   ОКПД: {record['okpd_code']}")
    okpd_display = record['okpd_name'][:70] + '...' if record['okpd_name'] and len(record['okpd_name']) > 70 else record['okpd_name']
    print(f"   └─ {okpd_display}\n")

# Дополнительно: группировка по ТН ВЭД (один ТН ВЭД → несколько ОКПД)
grouped_result = {}
for record in result:
    tnved = record['tnved_code']
    if tnved not in grouped_result:
        grouped_result[tnved] = {
            'tnved_code': tnved,
            'tnved_name': record['tnved_name'],
            'okpd_codes': []
        }
    grouped_result[tnved]['okpd_codes'].append({
        'okpd_code': record['okpd_code'],
        'okpd_name': record['okpd_name']
    })

# Сохраняем сгруппированный вариант
grouped_output = 'tnved_okpd_grouped.json'
with open(grouped_output, 'w', encoding='utf-8') as f:
    json.dump(list(grouped_result.values()), f, ensure_ascii=False, indent=2)

print(f"{'='*80}")
print(f"✓ Также создан сгруппированный файл: {grouped_output}")
print(f"  (один код ТН ВЭД → массив соответствующих ОКПД)")
```

```python
import os

rename_map = {
    '2. Данные для решения кейса_данные.json': 'case_data.json',
    'tws_тнвэд_ТНВЭД.json': 'tws_tnved.json',
    'Банкоматы_объем_импорта_Россия_Лист1.json': 'atms_import_russia_value.json',
    'Банкоматы_объем_импорта_Россия_Лист2.json': 'atms_import_russia_weight.json',
    'Лифты_объем_импорта_Россия_Лист1.json': 'elevators_import_russia_value.json',
    'Лифты_объем_импорта_Россия_Лист2.json': 'elevators_import_russia_weight.json',
    'Парфюмерия_объем_импорта_Россия_Лист1.json': 'perfumery_import_russia_value.json',
    'Парфюмерия_объем_импорта_Россия_Лист2.json': 'perfumery_import_russia_weight.json',
    'страны_список_Sheet1.json': 'countries_list.json'
}

base_path = 'data/json/'

for old_name, new_name in rename_map.items():
    old_path = os.path.join(base_path, old_name)
    new_path = os.path.join(base_path, new_name)
    if os.path.exists(old_path):
        os.rename(old_path, new_path)
        print(f'{old_name} -> {new_name}')
```

```python
import json

products = {
    'elevators': {
        'value': 'elevators_import_russia_value.json',
        'weight': 'elevators_import_russia_weight.json'
    },
    'perfumery': {
        'value': 'perfumery_import_russia_value.json',
        'weight': 'perfumery_import_russia_weight.json'
    },
    'atms': {
        'value': 'atms_import_russia_value.json',
        'weight': 'atms_import_russia_weight.json'
    }
}

base_path = 'data/json/'
combined = {}

for product, files in products.items():
    combined[product] = {
        'value': json.load(open(base_path + files['value'], 'r', encoding='utf-8')),
        'weight': json.load(open(base_path + files['weight'], 'r', encoding='utf-8'))
    }

with open(base_path + 'import_statistics.json', 'w', encoding='utf-8') as f:
    json.dump(combined, f, ensure_ascii=False, indent=2)
```
